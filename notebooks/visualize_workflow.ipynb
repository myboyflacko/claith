# Claith implementation-review graph

This notebook builds the real workflow without starting agents or processing events. It visualizes the compiled LangGraph structure used by Claith.

From the repository root, install and register the project kernel once:

```bash
uv sync --group dev
uv run python -m ipykernel install --user --name claith --display-name "Python (claith)"
```

Then launch the notebook and select **Python (claith)** as its kernel:

```bash
uv run --with jupyter jupyter lab notebooks/visualize_workflow.ipynb
```

````python
from IPython.display import Image, display
from langgraph.checkpoint.memory import InMemorySaver

from claith.application.workflow import build_workflow
````

````python
workflow = build_workflow(InMemorySaver())
drawable_graph = workflow.get_graph()
mermaid = drawable_graph.draw_mermaid()
````

## Diagram

`draw_mermaid_png()` uses Mermaid.ink and therefore requires an internet connection. The Mermaid source remains available in the next cell if image rendering is unavailable.

````python
try:
    display(Image(drawable_graph.draw_mermaid_png()))
except Exception as error:
    print(f"PNG rendering unavailable: {error}")
    print(mermaid)
````
